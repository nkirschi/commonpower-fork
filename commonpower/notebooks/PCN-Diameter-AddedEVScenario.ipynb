# Action Space Diameter

The action space diameter represents the range of control the agent has.

#### Case 1: With EV, Without Heat Pump
The agent controls the battery (`ESS1`) and the EV charging (`EVSE1`).
For the EV Charging Range we use `1e3`,  the maximum absolute charging power (BigM constraint) from `components.py`

* Battery Range: `[-1.5, 1.5]` kW
* EV Charging Range: `[-1000, +1000]` kW
* Diameter:
    $d_1 =\sqrt{3^2 + 2000^2} =\sqrt{9 + 4\,000\,000} =\sqrt{4\,000\,009} \approx 2000.00$

#### Case 2: With EV and Heat Pump
The agent controls the battery, EV, and heat pump.

* Battery Range: `[-1.5, 1.5]` kW
* EV Charging Range: `[-1000, +1000]` kW
* Heat Pump Range: `[0, 5.0]` kW
* Diameter:
    $d_2 =\sqrt{3^2 + 2000^2 + 5^2} =\sqrt{9 + 4\,000\,000 + 25} =\sqrt{4\,000\,034} \approx 2000.01$
