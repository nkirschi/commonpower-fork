# Action Space Diameter

The action space diameter is the range of control the agent has in the given scenario.

#### Case 1: With EV, Without Heat Pump
The agent controls the battery (`ESS1`).

* Battery Range: `[-1.5, 1.5]` (kW)
* Diameter:
    $d_1 = \sqrt{(1.5 - (-1.5))^2} = 3$

#### Case 2: With EV and Heat Pump
The agent controls the battery (`ESS1`), and heat pump.

* Battery Range: `[-1.5, 1.5]` (kW)
* Heat Pump Range: `[0, 5.0]` (kW)
* Diameter:
    $d_2 = \sqrt{3^2 + 5^2} = \sqrt{34} \approx 5.83$
